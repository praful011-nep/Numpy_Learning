****Array Operations****

**Arithmetic Operations**

Performs Operations on corresponding elements of array 
Size of arrays must be same

```python
a1=np.array([1,2,3,4,5])
a2=np.array([6,7,8,9,10])
```

```python
#Sum
a1 + a2
```

```python
#Difference
a1 - a2
```

```python
#Product
a1 * a2
```

```python
#Division
a1 / a2
```

```python
#Floor Division
a1 // a2
```

```python
#Power
a1 ** a2
```
