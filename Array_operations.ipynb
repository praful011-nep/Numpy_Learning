```python
#Division
a1 / a2
```

```python
#Floor Division
a1 // a2
```

```python
#Power
a1 ** a2
```

**Deep and Shallow Copy**  
*Shallow Copy: When a arry is sliced  and performed operations in it sliced copy, no changes will occur in the original array as both are in seperate location in RAM*  
*Deep Copy: When a array is sliced and performed operations in it, the original array also changes as both variables are in the same location of RAM*

```python
a=np.arange(1,21)
a
```

```python
slice = a[:5]
slice = slice *10
slice
```

```python
a
```

```python
a[4:6] *10 
```

```python
a
```

```python
b = a.copy() #Shallow Copy
# b=a     #Deep Copy
b[0] = 99 
```

```python
b
```

```python
a
```

**Matrix Operation**  

**Matrix Multiplication**  
*In Matrix, * is used for multiplication with boradcasting. But for actual matrix multiplication @ is used.*  
*Matrix1* @ *Matrix2*     *OR* np.dot(*Matrix1,Matrix2*) 

```python
m=np.array([[1,2],[4,5]])
n=np.array([[7,8],[3,6]])
```

```python
print("m:",m,"\nn:",n)
```

```python
m * n #Broadcasting
```

```python
m @ n #Matrix Multiplication
```
